```python
missing_data = df.isnull().sum()
print(missing_data[missing_data > 0])
```

```python
df_clean = df.copy()
print("Cleaned dataset copy created!")
```

```python
df_clean["Customer Name"] = df_clean["Customer Name"].fillna("Unknown")
```

```python
df_clean["City"] = df_clean["City"].fillna("Unknown")
```

```python
df_clean["Discount"] = df_clean["Discount"].fillna(
    df_clean["Discount"].median()
)
```

```python
print("Missing values after cleaning:")
print(df_clean.isnull().sum())
```

```python
print("Number of duplicate rows:")
print(df_clean.duplicated().sum())
```

```python
df_clean = df_clean.drop_duplicates()
print("Duplicate rows removed successfully!")
```

```python
print("Duplicates remaining:")
print(df_clean.duplicated().sum())
print("Updated dataset shape:")
print(df_clean.shape)
```

```python
print(df_clean.dtypes)
```

```python
df_clean["Order Date"]=pd.to_datetime(df_clean["Order Date"],errors="coerce")
print(df_clean["Order Date"].dtype)
```

```python
print(df_clean[df_clean["Quantity"]<=0])
```

```python
print(df_clean[df_clean["Sales"]<0])
```

```python
print(df_clean[(df_clean["Discount"]<0)|(df_clean["Discount"]>1)])
```

```python
Q1=df_clean["Sales"].quantile(0.25)
Q3=df_clean["Sales"].quantile(0.75)
IQR=Q3-Q1
print("Q1:",Q1)
print("Q3:",Q3)
print("IQR:",IQR)
```

```python
lower_limit=Q1-1.5*IQR
upper_limit=Q3+1.5*IQR
print("Lower Limit:",lower_limit)
print("Upper Limit:",upper_limit)
```

```python
outliers=df_clean[(df_clean["Sales"]<lower_limit)|(df_clean["Sales"]>upper_limit)]
print("Number of outliers:",len(outliers))
print(outliers[["Order ID","Sales"]])
```

```python
plt.figure(figsize=(8,4))
plt.boxplot(df_clean["Sales"])
plt.title("Sales Outlier Detection")
plt.ylabel("Sales")
plt.show()
```

```python
correlation=df_clean[["Quantity","Unit Price","Discount","Sales","Profit"]].corr()
print(correlation)
```

```python
plt.figure(figsize=(8,6))
plt.imshow(correlation,cmap="coolwarm",vmin=-1,vmax=1)
plt.colorbar()
plt.xticks(range(len(correlation.columns)),correlation.columns,rotation=45)
plt.yticks(range(len(correlation.columns)),correlation.columns)
plt.title("Correlation Heatmap")
for i in range(len(correlation.columns)):
    for j in range(len(correlation.columns)):
        plt.text(j,i,round(correlation.iloc[i,j],2),ha="center",va="center")
plt.tight_layout()
plt.show()
```

```python
print("Quantity and Sales correlation:",df_clean["Quantity"].corr(df_clean["Sales"]))
```

```python
df_clean["Month"]=df_clean["Order Date"].dt.month
df_clean["Year"]=df_clean["Order Date"].dt.year
```

```python
monthly_sales=df_clean.groupby(["Year","Month"])["Sales"].sum()
print(monthly_sales)
```

```python
monthly_sales=df_clean.groupby(df_clean["Order Date"].dt.to_period("M"))["Sales"].sum()
plt.figure(figsize=(12,5))
plt.plot(monthly_sales.index.astype(str),monthly_sales.values,marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.grid()
plt.tight_layout()
plt.show()
```

```python
print("Highest sales month:",monthly_sales.idxmax())
print("Highest monthly sales:",monthly_sales.max())
```

```python
print("Lowest sales month:",monthly_sales.idxmin())
print("Lowest monthly sales:",monthly_sales.min())
```

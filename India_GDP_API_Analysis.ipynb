```python
import requests
import json
import pandas as pd
```

```python
url = "https://api.worldbank.org/v2/country/IND/indicator/NY.GDP.MKTP.CD?format=json&per_page=100"

response = requests.get(url)

print("Status Code:", response.status_code)
```

```python
data = response.json()

print(json.dumps(data, indent=4)[:2000])
```

```python
with open("India_GDP_Data.json", "w") as file:
    json.dump(data, file, indent=4)

print("JSON file saved successfully!")
```

```python
records = data[1]

df = pd.DataFrame([
    {
        "Year": item["date"],
        "Country": item["country"]["value"],
        "GDP_USD": item["value"]
    }
    for item in records
    if item["value"] is not None
])

df.head()
```

```python
df.to_csv("India_GDP_Data.csv", index=False)

print("CSV file saved successfully!")
```

```python
print("Total Sales:", df["Sales"].sum())
print("Total Profit:", df["Profit"].sum())
print("Average Sales:", df["Sales"].mean())
print("Total Orders:", df["Order ID"].nunique())
print("Total Quantity:", df["Quantity"].sum())
```

```python
report = df.groupby("Region").agg(
    Total_Sales=("Sales", "sum"),
    Total_Profit=("Profit", "sum"),
    Average_Sales=("Sales", "mean"),
    Total_Quantity=("Quantity", "sum")
)
print(report)
```

```python
sales = np.array(df["Sales"])
profit = np.array(df["Profit"])
quantity = np.array(df["Quantity"])
```

```python
print("Total Sales:", np.sum(sales))
print("Average Sales:", np.mean(sales))
print("Median Sales:", np.median(sales))
print("Maximum Sales:", np.max(sales))
print("Minimum Sales:", np.min(sales))
print("Standard Deviation:", np.std(sales))
```

```python
print("Total Profit:", np.sum(profit))
print("Average Profit:", np.mean(profit))
print("Highest Profit:", np.max(profit))
print("Lowest Profit:", np.min(profit))
```

```python
print("Total Quantity:", np.sum(quantity))
print("Average Quantity:", np.mean(quantity))
```

```python
total_sales = np.sum(sales)
total_profit = np.sum(profit)

profit_margin = (total_profit / total_sales) * 100

print("Profit Margin:", profit_margin, "%")
```

```python
region_sales = df.groupby("Region")["Sales"].sum()

plt.bar(region_sales.index, region_sales.values)
plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.show()
```

```python
category_sales = df.groupby("Category")["Sales"].sum()

plt.pie(category_sales.values,
        labels=category_sales.index,
        autopct="%1.1f%%")

plt.title("Sales by Category")
plt.show()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])

monthly_sales = df.groupby(
    df["Order Date"].dt.to_period("M")
)["Sales"].sum()

plt.plot(monthly_sales.index.astype(str), monthly_sales.values, marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.scatter(df["Sales"], df["Profit"])
plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()
```

```python
plt.figure()
plt.bar(region_sales.index, region_sales.values)
plt.title("Sales by Region")
plt.savefig("sales_by_region.png")
plt.show()
```
